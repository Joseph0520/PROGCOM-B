Todos los números divisibles por 6 del 1 al 60

```python
numeros_divisibles = []
for num in range(1, 61):
    if num % 6 == 0:
        numeros_divisibles.append(num)
```

Todos los nombres que empiecen por "A" de una lista de nombres

Todos los números primos entre 0 y 70

```python
def encontrar_primos(limite):
    # Lista para guardar los primos
    primos = []

    # Revisar cada número del 2 al límite
    for numero in range(2, limite + 1):
        # Suponemos que el número es primo
        es_primo = True

        # Revisar si es divisible por algún número anterior
        for divisor in range(2, numero):
            if numero % divisor == 0:
                # Si es divisible, no es primo
                es_primo = False
                break

        # Si no encontramos divisores, es primo
        if es_primo:
            primos.append(numero)

    return primos
```
