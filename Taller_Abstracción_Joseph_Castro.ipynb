Crea una clase Rectangulo con atributos base y altura. Agrega un método

```python
class Rectangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def area(self):
        return self.base * self.altura

r = Rectangulo(5, 3)
print(r.area())
```

Crea una clase Palabra con un atributo texto. Agrega un método
contar_vocales(self) que cuente las vocales.
Instancia la clase con distintas palabras y prueba el método.

```python
class Palabra:
    def __init__(self, texto):
        self.texto = texto.lower()

    def contar_vocales(self):
        return sum(1 for c in self.texto if c in "aeiou")

p = Palabra("Hola")
print(p.contar_vocales())
```

```python
public class Palabra {
    String texto;

    public Palabra(String texto) {
        this.texto = texto.toLowerCase();
    }

    public int contarVocales() {
        int contador = 0;
        for (char c : texto.toCharArray()) {
            if ("aeiou".indexOf(c) != -1) contador++;
        }
        return contador;
    }

    public static void main(String[] args) {
        Palabra p = new Palabra("Hola");
        System.out.println(p.contarVocales());
    }
}
```

Crea una clase Palindromo con un atributo palabra. Agrega un método
es_palindromo(self) que retorne True si lo es.
Prueba con varias palabras como "reconocer", "luz", "oso"

```python
class Palindromo:
    def __init__(self, palabra):
        self.palabra = palabra.lower()

    def es_palindromo(self):
        return self.palabra == self.palabra[::-1]

p = Palindromo("oso")
print(p.es_palindromo())
```

```python
public class Palindromo {
    String palabra;

    public Palindromo(String palabra) {
        this.palabra = palabra.toLowerCase();
    }

    public boolean esPalindromo() {
        return new StringBuilder(palabra).reverse().toString().equals(palabra);
    }

    public static void main(String[] args) {
        Palindromo p = new Palindromo("oso");
        System.out.println(p.esPalindromo());
    }
}
```

Crea una clase ListaCuadrados con un método generar(self, n) que retorne
una lista con los cuadrados de los primeros n números.
Prueba con diferentes valores de n.


```python
class ListaCuadrados:
    def generar(self, n):
        return [i**2 for i in range(1, n+1)]

l = ListaCuadrados()
print(l.generar(5))
```

```python
import java.util.ArrayList;

public class ListaCuadrados {
    public ArrayList<Integer> generar(int n) {
        ArrayList<Integer> lista = new ArrayList<>();
        for (int i = 1; i <= n; i++) {
            lista.add(i * i);
        }
        return lista;
    }

    public static void main(String[] args) {
        ListaCuadrados l = new ListaCuadrados();
        System.out.println(l.generar(5));
    }
}
```

Crea una clase Calculadora con un método operar(self, a, b, tipo) que realice
la operación indicada: suma, resta, multiplicación o división.
Instancia y prueba con varios casos

```python
class Calculadora:
    def operar(self, a, b, tipo):
        if tipo == "suma":
            return a + b
        elif tipo == "resta":
            return a - b
        elif tipo == "multiplicacion":
            return a * b
        elif tipo == "division":
            return a / b

c = Calculadora()
print(c.operar(4, 2, "suma"))
```

```python
public class Calculadora {
    public double operar(double a, double b, String tipo) {
        return switch (tipo) {
            case "suma" -> a + b;
            case "resta" -> a - b;
            case "multiplicacion" -> a * b;
            case "division" -> a / b;
            default -> 0;
        };
    }

    public static void main(String[] args) {
        Calculadora c = new Calculadora();
        System.out.println(c.operar(4, 2, "suma"));
    }
}
```

Crea una clase GeneradorContrasena con un método crear(self, nombre, año)
que retorne la contraseña uniendo el nombre con el año y un símbolo como @.
Prueba el método con diferentes nombres y años.

```python
class GeneradorContrasena:
    def crear(self, nombre, año):
        return f"{nombre}{año}@"

g = GeneradorContrasena()
print(g.crear("usuario", 2025))
```

```python
public class GeneradorContrasena {
    public String crear(String nombre, int anio) {
        return nombre + anio + "@";
    }

    public static void main(String[] args) {
        GeneradorContrasena g = new GeneradorContrasena();
        System.out.println(g.crear("usuario", 2025));
    }
}
```
