```python
import random
import csv

class Paciente:
    def __init__(self, nombre, oxigeno, vacunas, prueba_covid):
        self.nombre = nombre
        self.oxigeno = oxigeno
        self.vacunas = vacunas
        self.prueba_covid = prueba_covid
        self.resultado = ""

    def necesita_sala_covid(self):
        return (
            self.oxigeno < 90 or
            self.vacunas < 2 or
            self.prueba_covid == "positivo"
        )


class SistemaTriage:
    def __init__(self, capacidad_urgencias):
        self.capacidad_urgencias = capacidad_urgencias
        self.contador_urgencias = 0

    def clasificar(self, paciente):
        if paciente.necesita_sala_covid():
            paciente.resultado = "SALA COVID"
        elif self.contador_urgencias < self.capacidad_urgencias:
            paciente.resultado = "URGENCIAS"
            self.contador_urgencias += 1
        else:
            paciente.resultado = "ESPERA"


# 1. Generar archivo CSV con 50 pacientes aleatorios
def generar_datos_csv(nombre_archivo):
    nombres = [f"Paciente_{i+1}" for i in range(50)]
    with open(nombre_archivo, mode='w', newline='') as archivo:
        writer = csv.writer(archivo)
        writer.writerow(["Nombre", "Oxigeno", "Vacunas", "Prueba COVID"])
        for nombre in nombres:
            oxigeno = random.randint(85, 100)
            vacunas = random.randint(0, 3)
            prueba = random.choice(["positivo", "negativo"])
            writer.writerow([nombre, oxigeno, vacunas, prueba])


# 2. Leer archivo, clasificar y guardar resultados
def procesar_datos_entrada(nombre_entrada, nombre_salida):
    pacientes = []
    with open(nombre_entrada, mode='r') as archivo:
        reader = csv.DictReader(archivo)
        for fila in reader:
            paciente = Paciente(
                fila["Nombre"],
                int(fila["Oxigeno"]),
                int(fila["Vacunas"]),
                fila["Prueba COVID"]
            )
            pacientes.append(paciente)

    sistema = SistemaTriage(capacidad_urgencias=10)
    for paciente in pacientes:
        sistema.clasificar(paciente)

    with open(nombre_salida, mode='w', newline='') as archivo:
        writer = csv.writer(archivo)
        writer.writerow(["Nombre", "Oxigeno", "Vacunas", "Prueba COVID", "Destino"])
        for p in pacientes:
            writer.writerow([p.nombre, p.oxigeno, p.vacunas, p.prueba_covid, p.resultado])


# Ejecutar todo
generar_datos_csv("entrada_pacientes.csv")
procesar_datos_entrada("entrada_pacientes.csv", "resultado_clasificacion.csv")
print("✅ Archivos generados: 'entrada_pacientes.csv' y 'resultado_clasificacion.csv'")

```
