En Python

```python
class Animal:
    def hablar(self):
        print("Este es un animal")

class Perro(Animal):
    pass

perro = Perro()
perro.hablar()
```

```python
class Persona:
    def __init__(self, nombre):
        self.nombre = nombre

class Estudiante(Persona):
    pass

est = Estudiante("Ana")
print(est.nombre)
```

```python
class Vehiculo:
    def arrancar(self):
        print("Vehículo encendido")

class Moto(Vehiculo):
    def arrancar(self):
        print("Moto encendida")

moto = Moto()
moto.arrancar()
```

```python
class Empleado:
    def trabajar(self):
        print("Trabajando...")

class Gerente(Empleado):
    def trabajar(self):
        print("Gerente trabajando...")

class Obrero(Empleado):
    def trabajar(self):
        print("Obrero trabajando...")

g = Gerente()
o = Obrero()
g.trabajar()
o.trabajar()
```

```python
class Producto:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio

class Libro(Producto):
    pass

libro = Libro("Python Básico", 30.5)
print(f"{libro.nombre} cuesta ${libro.precio}")
```

```python
class Persona:
    def saludar(self):
        print("Hola")

class Profesor(Persona):
    def enseñar(self):
        print("Estoy enseñando")

prof = Profesor()
prof.saludar()
prof.enseñar()
```

```python
class Animal:
    def sonido(self):
        print("Sonido genérico")

class Gato(Animal):
    def sonido(self):
        super().sonido()
        print("Miau")

gato = Gato()
gato.sonido()
```

```python
class Computadora:
    def encender(self):
        print("Computadora encendida")

class Portatil(Computadora):
    pass

portatil = Portatil()
portatil.encender()
```

```python
class Cuenta:
    def __init__(self, saldo):
        self.saldo = saldo

class CuentaAhorro(Cuenta):
    pass

cuenta = CuentaAhorro(1500)
print(f"Saldo: ${cuenta.saldo}")
```

```python
class SerVivo:
    def respirar(self):
        print("Respirando...")

class Humano(SerVivo):
    pass

humano = Humano()
humano.respirar()
```

```python
class Animal {
    void hablar() {
        System.out.println("Este es un animal");
    }
}

class Perro extends Animal {
}

public class Main1 {
    public static void main(String[] args) {
        Perro perro = new Perro();
        perro.hablar();
    }
}
```

```python
class Persona {
    String nombre;

    Persona(String nombre) {
        this.nombre = nombre;
    }
}

class Estudiante extends Persona {
    Estudiante(String nombre) {
        super(nombre);
    }
}

public class Main2 {
    public static void main(String[] args) {
        Estudiante est = new Estudiante("Ana");
        System.out.println(est.nombre);
    }
}
```

```python
class Vehiculo {
    void arrancar() {
        System.out.println("Vehículo encendido");
    }
}

class Moto extends Vehiculo {
    @Override
    void arrancar() {
        System.out.println("Moto encendida");
    }
}

public class Main3 {
    public static void main(String[] args) {
        Moto moto = new Moto();
        moto.arrancar();
    }
}
```

```python
class Empleado {
    void trabajar() {
        System.out.println("Trabajando...");
    }
}

class Gerente extends Empleado {
    @Override
    void trabajar() {
        System.out.println("Gerente trabajando...");
    }
}

class Obrero extends Empleado {
    @Override
    void trabajar() {
        System.out.println("Obrero trabajando...");
    }
}

public class Main4 {
    public static void main(String[] args) {
        Gerente g = new Gerente();
        Obrero o = new Obrero();
        g.trabajar();
        o.trabajar();
    }
}
```

```python
class Producto {
    String nombre;
    double precio;

    Producto(String nombre, double precio) {
        this.nombre = nombre;
        this.precio = precio;
    }
}

class Libro extends Producto {
    Libro(String nombre, double precio) {
        super(nombre, precio);
    }
}

public class Main5 {
    public static void main(String[] args) {
        Libro libro = new Libro("Java Básico", 45.99);
        System.out.println(libro.nombre + " cuesta $" + libro.precio);
    }
}
```
