Escribe un programa que calcule el índice de masa corporal (IMC) y determine la categoría (bajo peso, normal, sobrepeso, obesidad). Al final, el programa debe imprimir: el nombre del usuario, su categoría y una breve descripción de lo que debe hacer de acuerdo a su categoría.

```python
#Con for
# Programa para calcular el Índice de Masa Corporal (IMC)

def calcular_imc(peso, altura):
    """Calcula el IMC dividiendo el peso (kg) entre la altura al cuadrado (m)"""
    return peso / (altura ** 2)

def determinar_categoria(imc):
    """Determina la categoría según el valor del IMC"""
    if imc < 18.5:
        return "bajo peso"
    elif imc < 25:
        return "normal"
    elif imc < 30:
        return "sobrepeso"
    else:
        return "obesidad"

def obtener_recomendacion(categoria):
    """Devuelve una recomendación según la categoría de IMC"""
    recomendaciones = {
        "bajo peso": "Consultar con un nutricionista para aumentar peso de forma saludable, incrementar el consumo de calorías y proteínas.",
        "normal": "Mantener hábitos alimenticios saludables y realizar actividad física regularmente para conservar su peso ideal.",
        "sobrepeso": "Reducir moderadamente la ingesta calórica, aumentar la actividad física y consultar con un profesional de la salud.",
        "obesidad": "Consultar con un médico y nutricionista para desarrollar un plan personalizado de alimentación y ejercicio."
    }
    return recomendaciones[categoria]

def main():
    # Solicitar datos al usuario
    nombre = input("Ingrese su nombre: ")
    peso = float(input("Ingrese su peso en kilogramos: "))
    altura = float(input("Ingrese su altura en metros: "))

    # Calcular IMC
    imc = calcular_imc(peso, altura)

    # Determinar categoría
    categoria = determinar_categoria(imc)

    # Obtener recomendación
    recomendacion = obtener_recomendacion(categoria)

    # Imprimir resultados usando un ciclo for
    resultados = [
        f"Nombre: {nombre}",
        f"IMC: {imc:.2f}",
        f"Categoría: {categoria}",
        f"Recomendación: {recomendacion}"
    ]

    print("\n--- RESULTADOS DEL CÁLCULO DE IMC ---")
    for i in range(len(resultados)):
        print(resultados[i])
    print("-------------------------------------")

# Ejecutar el programa
if __name__ == "__main__":
    main()
```

12.Desarrolla un programa que simule una calculadora simple (+, -, *, /) con dos números.

```python
# Versión con ciclo FOR
def calculadora_for():
    print("\n=== CALCULADORA CON CICLO FOR ===")

    # Solicitar dos números
    num1 = float(input("Ingrese el primer número: "))
    num2 = float(input("Ingrese el segundo número: "))

    # Lista de operaciones disponibles
    operaciones = [
        ("+", "suma"),
        ("-", "resta"),
        ("*", "multiplicación"),
        ("/", "división", lambda x, y: x / y if y != 0 else "Error: División por cero")
    ]

    # Mostrar menú y realizar cálculos usando for
    print("\nResultados:")
    for simbolo, nombre, operacion in operaciones:
        resultado = operacion(num1, num2)
        print(f"{num1} {simbolo} {num2} = {resultado}")

    print("==============================")
```

```python
# Versión con ciclo WHILE
def calculadora_while():
    print("\n=== CALCULADORA CON CICLO WHILE ===")

    # Solicitar dos números
    num1 = float(input("Ingrese el primer número: "))
    num2 = float(input("Ingrese el segundo número: "))

    # Lista de operaciones disponibles
    operaciones = [
        ("+", "suma"),
        ("-", "resta"),
        ("*", "multiplicación"),
        ("/", "división", lambda x, y: x / y if y != 0 else "Error: División por cero")
    ]

    # Mostrar menú y realizar cálculos usando while
    print("\nResultados:")
    i = 0
    while i < len(operaciones):
        simbolo, nombre, operacion = operaciones[i]
        resultado = operacion(num1, num2)
        print(f"{num1} {simbolo} {num2} = {resultado}")
        i += 1

    print("==============================")
```

Crea un programa que determine la calificación en letras (A, B, C, D, F) según una nota numérica (0-100)

```python
def obtener_calificacion(nota):
    """
    Determina la calificación en letras según la nota numérica
    A: 90-100
    B: 80-89
    C: 70-79
    D: 60-69
    F: 0-59
    """
    if nota >= 90 and nota <= 100:
        return "A"
    elif nota >= 80 and nota < 90:
        return "B"
    elif nota >= 70 and nota < 80:
        return "C"
    elif nota >= 60 and nota < 70:
        return "D"
    elif nota >= 0 and nota < 60:
        return "F"
    else:
        return "Nota inválida (debe estar entre 0 y 100)"

def obtener_descripcion(letra):
    """Devuelve una descripción según la calificación en letra"""
    descripciones = {
        "A": "Excelente desempeño",
        "B": "Buen desempeño",
        "C": "Desempeño promedio",
        "D": "Desempeño por debajo del promedio",
        "F": "Desempeño insuficiente"
    }
    return descripciones.get(letra, "")


    num_estudiantes = int(input("Ingrese el número de estudiantes: "))

    # Lista para almacenar los datos de los estudiantes
    estudiantes = []

    for i in range(num_estudiantes):
        print(f"\nEstudiante #{i+1}")
        nombre = input("Nombre: ")
        nota = float(input("Nota (0-100): "))
        letra = obtener_calificacion(nota)
        descripcion = obtener_descripcion(letra)
        estudiantes.append((nombre, nota, letra, descripcion))

    print("\n--- RESULTADOS ---")
    for nombre, nota, letra, descripcion in estudiantes:
        print(f"Estudiante: {nombre}")
        print(f"Nota: {nota}")
        print(f"Calificación: {letra}")
        print(f"Descripción: {descripcion}")
        print("-----------------")

    print("=================================")

```

```python
def calificacion_while():
    """Versión con ciclo WHILE"""
    print("\n=== CALIFICACIÓN EN LETRAS (WHILE) ===")

    # Solicitar notas al usuario
    num_estudiantes = int(input("Ingrese el número de estudiantes: "))

    # Lista para almacenar los datos de los estudiantes
    estudiantes = []

    # Recopilar datos
    contador = 0
    while contador < num_estudiantes:
        print(f"\nEstudiante #{contador+1}")
        nombre = input("Nombre: ")
        nota = float(input("Nota (0-100): "))
        letra = obtener_calificacion(nota)
        descripcion = obtener_descripcion(letra)
        estudiantes.append((nombre, nota, letra, descripcion))
        contador += 1

    # Mostrar resultados
    print("\n--- RESULTADOS ---")
    contador = 0
    while contador < len(estudiantes):
        nombre, nota, letra, descripcion = estudiantes[contador]
        print(f"Estudiante: {nombre}")
        print(f"Nota: {nota}")
        print(f"Calificación: {letra}")
        print(f"Descripción: {descripcion}")
        print("-----------------")
        contador += 1

    print("=================================")
```

Escribe un programa que determine el día de la semana según un número (1-7).

```python
def obtener_dia_semana(numero):
    """
    Convierte un número (1-7) al día de la semana correspondiente
    """
    dias = {
        1: "Lunes",
        2: "Martes",
        3: "Miércoles",
        4: "Jueves",
        5: "Viernes",
        6: "Sábado",
        7: "Domingo"
    }
    return dias.get(numero, "Número inválido (debe estar entre 1 y 7)")

def version_for():
    """Versión utilizando ciclo FOR"""
    print("\n=== DÍA DE LA SEMANA (VERSION FOR) ===")

    consultas = []

    num_consultas = int(input("¿Cuántas consultas desea realizar? "))

    for i in range(num_consultas):
        print(f"\nConsulta #{i+1}")
        try:
            numero = int(input("Ingrese un número (1-7): "))
            dia = obtener_dia_semana(numero)
            consultas.append((numero, dia))
        except ValueError:
            print("Error: Debe ingresar un número entero.")
            consultas.append((0, "Error: Entrada inválida"))

    print("\n--- RESULTADOS ---")
    for numero, dia in consultas:
        print(f"Número {numero} → {dia}")

    print("==============================")
```

Escribe un programa que calcule la suma de números naturales hasta n

```python
def sumar_naturales():
    suma = 0
    while True:
        try:
            n = int(input("Ingrese un número natural (o un número negativo para terminar): "))
            if n < 0:
                break
            suma += n
        except ValueError:
            print("Entrada inválida. Por favor, ingrese un número entero.")
    print(f"La suma de los números ingresados es: {suma}")

sumar_naturales()
```

Desarrolla un programa que calcule el factorial de un número. Recuerda que el factorial de 0 es 1 y no existen factoriales negativos

```python
def cf():
   while True:
       try:
           numero = int(input("Ingrese un número para calcular su factorial: "))
           if numero < 0:
               print("No existen factoriales de números negativos")
               continue
           factorial = 1
           for i in range(1, numero + 1):
               factorial *= i
           print(f"El factorial de {numero} es: {factorial}")
           break
       except ValueError:
           print("Por favor ingrese un número entero válido")
cf()
```

```python
# Pedimos al usuario que ingrese un número
print("Programa para calcular el factorial de un número")
numero = int(input("Ingrese un número entero no negativo: "))

# Verificamos que el número sea no negativo
if numero < 0:
    print("Error: No existe el factorial de números negativos.")
else:
    # Inicializamos el factorial como 1 (importante para el caso de 0! = 1)
    factorial = 1

    # Calculamos el factorial usando un ciclo for
    for i in range(1, numero + 1):
        factorial = factorial * i
        print(f"Multiplicando por {i}: {factorial}")

    # Mostramos el resultado
    print(f"\nEl factorial de {numero} es: {factorial}")

    # Mostramos la expresión matemática del factorial
    expresion = "1"
    if numero > 0:
        expresion = " × ".join(str(i) for i in range(1, numero + 1))

    print(f"{numero}! = {expresion} = {factorial}")
```

Crea un programa que muestre los primeros n números de la serie Fibonacci.

```python
def fibonacci():
   while True:
       try:
           n = int(input("¿Cuántos números de la serie Fibonacci desea ver? "))

           if n <= 0:
               print("Por favor ingrese un número positivo")
               continue
           a, b = 0, 1
           for i in range(n):
               print(a, end=" ")
               a, b = b, a + b
           print()
           break

       except ValueError:
           print("Por favor ingrese un número entero válido")

fibonacci()
```

Escribe un programa que adivine entre un conjunto de diez colores definidos por el usuario, con un máximo de 10 intentos

```python
import random

def juego_adivinar_colores():

  colores = ['amarillo', 'azul', 'rojo', 'verde', 'naranja',
              'morado', 'rosa', 'blanco', 'negro', 'gris']

   # Seleccionar un color al azar
   color_secreto = random.choice(colores)
   intentos = 0

   print("\n¡Bienvenido al juego de adivinar colores!")
   print("Estoy pensando en uno de estos colores:")
   print(", ".join(colores))
   print("\nTienes 10 intentos para adivinar")

   while intentos < 10:
       intento = input("\nIngresa un color: ").lower()
       intentos += 1

       if intento not in colores:
           print("Ese color no está en la lista. Intenta con uno de los colores mostrados")
           continue

       if intento == color_secreto:
           print(f"\n¡Felicitaciones! Adivinaste en {intentos} intentos")
           print(f"El color era: {color_secreto}")
           return
       else:
           intentos_restantes = 10 - intentos
           print(f"Color incorrecto. Te quedan {intentos_restantes} intentos")

   print(f"\nSe acabaron los intentos. El color era: {color_secreto}")
```

```python
import random

def juego_colores():
    print("¡Bienvenido al juego de adivinar colores!")
    print("Por favor, introduce diez colores diferentes.")


    colores = []

    # Solicitar al usuario que ingrese diez colores
    for i in range(10):
        color = input(f"Color {i+1}: ").strip().lower()
        while color in colores:
            print("Este color ya está en la lista. Introduce uno diferente.")
            color = input(f"Color {i+1}: ").strip().lower()
        colores.append(color)

    print("\nHas definido estos colores:")
    for i, color in enumerate(colores):
        print(f"{i+1}. {color}")

    # Seleccionar un color aleatorio
    color_secreto = random.choice(colores)

    print("\nHe seleccionado un color aleatorio de tu lista. ¡Intenta adivinarlo!")
    print("Tienes un máximo de 10 intentos.")

    # Bucle de intentos con for (máximo 10)
    for intento in range(1, 11):
        respuesta = input(f"\nIntento {intento}: ").strip().lower()

        if respuesta == color_secreto:
            print(f"¡Correcto! El color era {color_secreto}.")
            print(f"Has adivinado en {intento} {'intento' if intento == 1 else 'intentos'}.")
            return
        else:
            if intento < 10:
                print("Incorrecto. Intenta de nuevo.")
            else:
                print(f"Se acabaron los intentos. El color era {color_secreto}.")

# Ejecutar el juego
if __name__ == "__main__":
    juego_colores()
```
